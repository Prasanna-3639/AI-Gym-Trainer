```python

from google.colab import userdata

API_KEY = userdata.get("GEMINI_API_KEY")
```

```python
import google.genai as genai

client = genai.Client(api_key=API_KEY)
```

```python
from google.genai import types

config = types.GenerateContentConfig(temperature=0.1)
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="suggest five good movies for a kid.",
    config=config
)

print(response.text)
```

```python
# Temperature and top_p

config = types.GenerateContentConfig(
    temperature=0.9,
    top_p=0.7
)
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="suggest five good names for a new restaurant.",
    config=config
)

print(response.text)
```

```python
# temp, top_p, and top_k

config = types.GenerateContentConfig(
    temperature=0.9,
    top_p=0.7,
    top_k=5
)
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="suggest five good names for a new restaurant.",
    config=config
)

print(response.text)
```

```python
# temp, top_p, and top_k

config = types.GenerateContentConfig(
    temperature=0.9,
    top_p=0.7,
    top_k=5,
    max_output_tokens=1500
)
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="suggest five good names for a new restaurant.",
    config=config
)

print(response.text)
```

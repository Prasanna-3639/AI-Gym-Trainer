```python
from google.colab import userdata
from google import genai

API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=API_KEY)
```

```python
def build_prompt(role, query):
    role = role.lower()

    if role == "tutor":
        return f"You are a helpful tutor. Explain the following concept clearly:\n{query}"

    elif role == "career-counselor":
        return f"You are a career counselor. Advise a student interested in:\n{query}"

    elif role == "assistant":
        return f"You are a smart assistant. Help with the following request:\n{query}"

    else:
        return f"You are a general-purpose AI. Respond to:\n{query}"
```

```python
# Create prompt
prompt = build_prompt("tutor", "Explain Python inheritance")

# Generate response
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt
)

# Print response
print(response.text)
```

```python
# Generate response using Gemini
def generate_response(role, query):
    prompt = build_prompt(role, query)

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    return response.text
```

```python
response = generate_response(
    "career_counselor",
    "I am interested in becoming a Generative AI Engineer. What skills should I learn?"
)

print(response)
```

```python
from google.colab import userdata
from google import genai
API_KEY=userdata.get("GEMINI_API_KEY")
client= genai.Client(api_key=API_KEY)
```

```python
def generate_creative_text(style, topic):
    style = style.lower()

    if style == "story":
        prompt = f"Write a short story about: {topic}"

    elif style == "poem":
        prompt = f"Write a poem on the theme: {topic}"

    elif style == "blog":
        prompt = f"Write a blog post about: {topic}"

    elif style == "script":
        prompt = f"Write a short script based on: {topic}"

    else:
        prompt = f"Write creatively about: {topic}"

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    return response.text


# Example usage
print(generate_creative_text("story", "an AI that learns emotions"))

print(generate_creative_text("poem", "the beauty of Hyderabad in monsoon"))

print(generate_creative_text("blog", "how generative AI is transforming education"))
```
